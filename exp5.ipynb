```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('img.jpg')

img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
# Averaging Filter
avg = cv2.blur(img, (5, 5))

# Weighted Averaging Filter
kernel = np.array([[1, 2, 1],
                   [2, 4, 2],
                   [1, 2, 1]], dtype=np.float32) / 16

weighted = cv2.filter2D(img, -1, kernel)

# Gaussian Filter
gaussian = cv2.GaussianBlur(img, (5, 5), 0)

# Median Filter
median = cv2.medianBlur(img, 5)

# Display results
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(avg)
plt.title("Averaging Filter")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(weighted)
plt.title("Weighted Averaging")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(gaussian)
plt.title("Gaussian Filter")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(median)
plt.title("Median Filter")
plt.axis("off")

plt.show()
```

```python
# Laplacian Sharpening using Kernel
kernel = np.array([[0, -1, 0],
                   [-1, 5, -1],
                   [0, -1, 0]])

lap_kernel = cv2.filter2D(img, -1, kernel)

# Convert image to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)

# Laplacian Operator
lap = cv2.Laplacian(gray, cv2.CV_64F)

lap = cv2.convertScaleAbs(lap)

# Display results
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(lap_kernel)
plt.title("Laplacian Kernel")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(lap, cmap='gray')
plt.title("Laplacian Operator")
plt.axis("off")

plt.show()
```

```python
plt.figure(figsize=(15, 10))

images = [
    img, avg, weighted, gaussian,
    median, lap_kernel, lap
]

titles = [
    "Original Image",
    "Averaging",
    "Weighted Averaging",
    "Gaussian",
    "Median",
    "Laplacian Kernel",
    "Laplacian Operator"
]

for i in range(7):
    plt.subplot(3, 3, i + 1)

    if i == 6:
        plt.imshow(images[i], cmap='gray')
    else:
        plt.imshow(images[i])

    plt.title(titles[i])
    plt.axis("off")

plt.tight_layout()
plt.show()
```
